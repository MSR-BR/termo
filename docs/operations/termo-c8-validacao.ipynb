## tl;dr
Validação reproduzível das participações e custos por conversão usados na C8.

## Context & Methods
Fonte: interface do Google Ads, conta MSR-BR, campanha TERMO - Search - BR - PT, período de 5 de agosto a 3 de setembro de 2026. As participações por rede são arredondadas pelo Google Ads.

### Key Assumptions
O total de 141 conversões e €99,13 é distribuído pelas participações exibidas para cada rede.

## Data

```python
total_cost = 99.13
total_conversions = 141
networks = {
    'Google Pesquisa': {'conversion_share': 0.582, 'cost_share': 0.757},
    'Parceiros de pesquisa': {'conversion_share': 0.418, 'cost_share': 0.243},
}
```

## Results

```python
results = {}
for name, shares in networks.items():
    conversions = total_conversions * shares['conversion_share']
    cost = total_cost * shares['cost_share']
    results[name] = {
        'conversions_estimated': round(conversions, 2),
        'cost_estimated_eur': round(cost, 2),
        'cost_per_conversion_eur': round(cost / conversions, 2),
    }
results
```

```python
assert round(sum(v['conversion_share'] for v in networks.values()), 6) == 1
assert round(sum(v['cost_share'] for v in networks.values()), 6) == 1
assert results['Parceiros de pesquisa']['cost_per_conversion_eur'] < results['Google Pesquisa']['cost_per_conversion_eur']
results
```

## Takeaways
Parceiros de pesquisa usam uma parcela menor do custo do que das conversões e apresentam custo por conversão estimado inferior. Os valores são aproximações, não medições brutas por rede.
